# **Выбросы**

```python
def outliers1(data):
    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    up = q3 + 1.5 * iqr
    return data[(data < low) | (data > up)]

out = emp.select_dtypes(['int', 'float']).apply(outliers1)
out.info()
```

По медоту 1.5 * iqr в нашем датасете есть выбросы в следующих столбцах: Academic Reputation, Employer Reputation, Faculty Student, International Students, Employment Outcomes, Sustainability

Удалим выбросы из датасета:

```python
q1 = emp['Academic Reputation'].quantile(0.25)
q3 = emp['Academic Reputation'].quantile(0.75)
iqr = q3 - q1
low1 = q1 - 1.5 * iqr
up1 = q3 + 1.5 * iqr
emp_clean = emp[(emp['Academic Reputation'] >= low1) & (emp['Academic Reputation'] <= up1)]
```

```python
q1 = emp['Employer Reputation'].quantile(0.25)
q3 = emp['Employer Reputation'].quantile(0.75)
iqr = q3 - q1
low1 = q1 - 1.5 * iqr
up1 = q3 + 1.5 * iqr
emp_clean1 = emp_clean[(emp_clean['Employer Reputation'] >= low1) & (emp_clean['Employer Reputation'] <= up1)]
```

```python
q1 = emp['Faculty Student'].quantile(0.25)
q3 = emp['Faculty Student'].quantile(0.75)
iqr = q3 - q1
low1 = q1 - 1.5 * iqr
up1 = q3 + 1.5 * iqr
emp_clean2 = emp_clean1[(emp_clean1['Faculty Student'] >= low1) & (emp_clean1['Faculty Student'] <= up1)]
```

```python
q1 = emp['International Students'].quantile(0.25)
q3 = emp['International Students'].quantile(0.75)
iqr = q3 - q1
low1 = q1 - 1.5 * iqr
up1 = q3 + 1.5 * iqr
emp_clean3 = emp_clean2[(emp_clean2['International Students'] >= low1) & (emp_clean2['International Students'] <= up1)]
```

```python
q1 = emp['Employment Outcomes'].quantile(0.25)
q3 = emp['Employment Outcomes'].quantile(0.75)
iqr = q3 - q1
low1 = q1 - 1.5 * iqr
up1 = q3 + 1.5 * iqr
emp_clean4 = emp_clean3[(emp_clean3['Employment Outcomes'] >= low1) & (emp_clean3['Employment Outcomes'] <= up1)]
```

```python
q1 = emp['Sustainability'].quantile(0.25)
q3 = emp['Sustainability'].quantile(0.75)
iqr = q3 - q1
low1 = q1 - 1.5 * iqr
up1 = q3 + 1.5 * iqr
emp_clean5 = emp_clean4[(emp_clean4['Sustainability'] >= low1) & (emp_clean4['Sustainability'] <= up1)]

emp_clean5
```

# **ГИПОТЕЗЫ**

1. **Хи-квадрат :**

Мы предполагаем, что страна, в которой находится университет, и его размер не связаны.

   
H₀: Размер университета не зависит от страны.

H₁: Эти переменные зависимы.


```python
df = pd.crosstab(emp_clean5['Location'], emp_clean5['Size'])
df
```

```python
from scipy.stats import chi2_contingency

res = chi2_contingency(df.values)
res
```

```python
print(res[1] > 0.1)
print(res[1] > 0.05)
print(res[1] > 0.01)
```

Основная гипотеза **отвергается** при любом показателе значимости. Таким образом, размер университетов зависит от страны. Возможно, это связано с тем, что разные государства имеют разные образовательные системы, экономические и исторические особенности.

2. **Одновыборочный t-тест:**

Наш датасет содержит рейтинг топовых университетов, поэтому мы предполагаем, что средний показатель академической репутации университетов равен 95.

H₀: Средний показатель академической репутации университетов равен 95.
    
H₁: Средний показатель академической репутации < 95.


```python
from scipy import stats

result = stats.ttest_1samp(a=emp_clean5['Academic Reputation'],
                           popmean=95,
                           alternative='less')
result
```

```python
print(result.pvalue > 0.1) #уровень значимости 10%
print(result.pvalue > 0.05) #уровень значимости 5%
print(result.pvalue > 0.01) #уровень значимости 1%
```

p_value < α -> Основная гипотеза **отвергается** при любом показателе значимости. Несмотря на то что наш датасет включает топовые вузы, среднее значение академической репутации ниже 95. Это показывает, что академическая репутация не является главным критерием для оценки уровня вуза.

3. **Двухвыборочный t-тест:**

США и Великобритания - англоговорящие страны, поэтому мы предполагаем, что средние значения цитируемости университетов из США и Великобритании не различаются

H₀: Средние значения цитируемости (Citations per Faculty) университетов из США и Великобритании не различаются.

H₁: Средние значения различаются.

```python
USA = emp_clean5[(emp_clean5['Location'] == 'US')]
UK = emp_clean5[(emp_clean5['Location'] == 'UK')]
```

```python
USA['Employer Reputation'].var() == UK['Employer Reputation'].var()
```

То есть дисперсии неравны

```python
from scipy.stats import ttest_ind as ttest_ind

a = np.array(USA['Employer Reputation'])
b = np.array(UK['Employer Reputation'])
res = ttest_ind(a, b, equal_var = False)
res
```

```python
print(res.pvalue > 0.1) #уровень значимости 10%
print(res.pvalue > 0.05) #уровень значимости 5%
print(res.pvalue > 0.01) #уровень значимости 1%
```

При уровнях значимости 1% и 5% основная гипотеза **не отвергается**, то есть средние значения цитируемости университетов из США и Великобритании не различаются. Это можно объяснить тем, что в университетах обеих стран цитируемая литература представлена на английском языке
При уровне значимости 10% основная гипотеза отвергается, то есть цитируемость в США и Великобритании различаются. На это могут влиять многие факторы, например, цитируемость университета зависит не от страны его нахождения, а от престижности


# Линейная регрессия

В качестве целевой переменной мы выбрали академическую репутацию универсетита (Academic Reputation), в качестве независимых переменных мы выбрали Employer Reputation, International Research Network, Sustainability, так как именно с этими признаками Academic Reputation имеет наибольшую корреляцию

```python
import statsmodels.api as sm

X1 = emp_clean5[['Employer Reputation', 'International Research Network', 'Sustainability']]
y = emp_clean5['Academic Reputation']

X1 = sm.add_constant(X1)

model1 = sm.OLS(y, X1).fit()
```
